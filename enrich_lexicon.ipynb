```python
!wget 'https://github.com/timarkh/uniparser-grammar-eastern-armenian/raw/refs/heads/master/hye-lexemes-A.txt'
!wget 'https://github.com/timarkh/uniparser-grammar-eastern-armenian/raw/refs/heads/master/hye-lexemes-ADV.txt'
!wget 'https://github.com/timarkh/uniparser-grammar-eastern-armenian/raw/refs/heads/master/hye-lexemes-N.txt'
!wget 'https://github.com/timarkh/uniparser-grammar-eastern-armenian/raw/refs/heads/master/hye-lexemes-V.txt'
!wget 'https://github.com/timarkh/uniparser-grammar-eastern-armenian/raw/refs/heads/master/hye-lexemes-rest.txt'
```

```python
dict_reprs = {}
for pos in ['A', 'ADV', 'N', 'V', 'rest']:
    print(pos)
    with open('hye-lexemes-' + pos + '.txt', encoding='utf8') as f:
        text = f.read().strip('\n').replace('\ufeff', '\n\n')
        entry_list = [
            {j.split(': ')[0]: j.split(': ')[1] \
             for j in i.split('\n ') if j != ''} \
             for i in text.split('\n\n-lexeme') if i != ''
             ]
        dict_reprs[pos] = {e['lex']: e for e in entry_list}
        print(len(dict_reprs[pos]))
        print(dict_reprs.keys())
        print()
```

### Add Russian translations

```python
from collections import defaultdict as dd
from tqdm import tqdm
```

```python
for pos in ['A', 'ADV', 'N', 'V', 'rest']:
    for e in dict_reprs[pos]:
        dict_reprs[pos][e]['trans_ru'] = ''
```

```python
dict_reprs['V']['ֆրթացնել']
```

```python
def arm_lower(text):
    return text.replace('ԵՎ', 'ԵՒ').lower()
```

```python
ct = dd(int)
OOV = 0
for _, row in tqdm(df.iterrows()):
    current_pos = row['pos']
    lemma_lower = arm_lower(row['lemma'])
    if current_pos in dict_reprs:
        if lemma_lower in dict_reprs[current_pos]:
            dict_reprs[current_pos][lemma_lower]['trans_ru'] = \
            str(row['sense_short']).lower()
            ct[current_pos] += 1
        else:
            # lemma not in parser's dict for this POS
            # unknown POS not handled here
            OOV += 1
```

```python
print(ct)
print(OOV) # it is possible to later attend to it
```

### Create specific Nor-Nakhichevan lexemes

Here I rely on files previously automaticlly compiled from the dictionary and then altered manually with regexp

```python
local_dict_reprs = {}
for pos in ['A', 'ADV', 'N', 'V', 'rest']:
    print(pos)
    with open('hynn-lexemes-' + pos + '.txt', encoding='utf8') as f:
        text = f.read().strip('\n').replace('\ufeff', '\n\n')
        entry_list = [
            {j.split(': ')[0]: j.split(': ')[1] \
             for j in i.split('\n ') if j != ''} \
             for i in text.split('\n\n-lexeme') if i != ''
             ]
        local_dict_reprs[pos] = {e['lex']: e for e in entry_list}
        print(len(local_dict_reprs[pos]))
        print(local_dict_reprs.keys())
        print()
```

```python
for pos in ['A', 'ADV', 'N', 'V', 'rest']:
    print(len(set(dict_reprs[pos].keys()).intersection(set(local_dict_reprs[pos].keys()))))
```

```python
for pos in ['A', 'ADV', 'N', 'V', 'rest']:
    for e in dict_reprs[pos]:
        dict_reprs[pos][e]['lemma_type'] = 'literary'
```

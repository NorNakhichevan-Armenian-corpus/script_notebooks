### Compare the two parsers' quantitative results

Here ambiguity is not taken into account well enough

```python
NNA_corpus_wordlist = df['token'].to_list()
print(len(NNA_corpus_wordlist))
print(NNA_corpus_wordlist[:5])
```

```python
with open('EA_parser_unanalyzed.txt', encoding='utf8') as f:
    EA_unanalyzed = [i for i in f.read().split('\n') if i != '']
print(f'EA_unanalyzed: {len(EA_unanalyzed)}')
print(EA_unanalyzed[:5])
print()

with open('NNA_parser_unanalyzed.txt', encoding='utf8') as f:
    NNA_unanalyzed = [i for i in f.read().split('\n') if i != '']
print(f'NNA_unanalyzed: {len(NNA_unanalyzed)}')
print(NNA_unanalyzed[:5])
```

```python
NNA_corpus_wordlist = set(NNA_corpus_wordlist)
EA_unanalyzed = set(EA_unanalyzed)
NNA_unanalyzed = set(NNA_unanalyzed)

EA_analyzed = NNA_corpus_wordlist.difference(EA_unanalyzed)
print(f'EA_analyzed: {len(EA_analyzed)}')
print()

NNA_analyzed = NNA_corpus_wordlist.difference(NNA_unanalyzed)
print(f'NNA_analyzed: {len(NNA_analyzed)}')
```

The number of tokens types for which the Nor-Nakhichevan parser did not do better than the baseline Eastern Armenian parser

```python
same_result = NNA_unanalyzed.intersection(EA_unanalyzed)
print(len(same_result))
print(list(same_result)[:10])
```

The number of tokens for which the Nor-Nakhichevan parser avoided Eastern parses (i.e., the NNA parser did not recognize them, while the EA parser did)

```python
# the difference between two sets includes the elements that are
# present in the first set but not present in the second one

only_EA_parses = EA_analyzed.difference(NNA_analyzed)
print(len(only_EA_parses))
print(list(only_EA_parses)[:10])
```

The number of tokens for which the Nor-Nakhichevan parser showed improvement (i.e., the NNA parser recognized them, while the EA parser did not)

```python
only_NNA_parses = NNA_analyzed.difference(EA_analyzed)
print(len(only_NNA_parses))
print(list(only_NNA_parses)[:10])
```

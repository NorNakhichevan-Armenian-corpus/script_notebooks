### Xatamyan files: .docx format

```python
path = 'Xatamyan1_converted_from_pdf.docx'
document = Document(path)
print(sum(len(para.text) for para in document.paragraphs))
print(document.paragraphs[0].text)
```

```python
def replace_in_paragraph(paragraph):
    for run in paragraph.runs:
        # inplace substitutions
        run.text = transpose_text(decode_every, run.text)

def replace_in_table(table):
    for row in table.rows:
        for cell in row.cells:
            for paragraph in cell.paragraphs:
                replace_in_paragraph(paragraph)
            for nested_table in cell.tables:
                replace_in_table(nested_table)

def replace_in_container(container):
    for paragraph in container.paragraphs:
        replace_in_paragraph(paragraph)
    for table in container.tables:
        replace_in_table(table)
```

```python
# main body of the text in the document
replace_in_container(document)

# other sections
for section in document.sections:
    replace_in_container(section.header)
    replace_in_container(section.footer)
```

```python
print(document.paragraphs[0].text)
print(sum(len(para.text) for para in document.paragraphs))
document.save(path[:-5] + '_decoded.docx')
```

Punctuation symbols, above all '»', need to be treated. This can be done semi-automatically, searching with the regexp '«.+[а-яА-ЯёЁԱ-Ֆ]{2,}ե' (sometimes this symbol was used after Armenian text, which cannot be easily noticed, only after all-capital-letter words). Also, in some cases the quotation marks written as ',,' (two commas) and with the special symbol '’' (that had to be decoded) for consistency are replaced with '„' and '”' (or '»'), respectively (in one case "ինձի մեն պակաս չէ" '’' was deleted). This mostly concerns the first book.

Semi-automatically we also correct the spaces sometimes inserted before punctuation marks ':', ',' and '՝'; the cases when a word is split by '- ' (using '\w- \w') and when hyphenated words are split by linebreak ('([ա-ևԱ-Ֆ])- \r\n'). Although we could not be 100% sure of what the punctuation should be (especially because rarely a there is a "true" hyphen that is intended by the author at the linebreak). We automatically replace the punctuation marks combination ',–' without a space with ', –' (this mostly concerns the second book)

```python
# we run these manually
def clear_text(text):
    interim = text
    interim = re.sub(' {2,}', ' ', interim)
    interim = re.sub('\t', ' ', interim)
    interim = re.sub('\r\n', '\n', interim)
    interim = re.sub('( ?\n){2,}', ' \n\n', interim)
    interim = re.sub('-\n', '', interim)
    return interim
```

This drawback remains in the texts: newline symbols when in fact the paragraph continues. It is hard to treat automatically because, e.g., in poems the line can end with non-verdjaket symbols. So in general **the paragraph structure of the document should not be trusted**

### Popovyan file: punctuation treatment

Firstly, one needs to preserve the true Cyrillic symbols, i.e., those that encodde Russian words, otherwise they will be replaced. This is probably best achieved by manually inserting special boundary marks and later backwards-replacement. (Alternatively, such code-switching phenomena can be deleted altogether)

```python
print('– Здравствуйте, мать, – ասաց մայոռը')
print('%%%%%– Здравствуйте, мать%%%%%, – ասաց մայոռը')
```

Then the replacements above need to be implemented

```python
filename = 'Popovyan_copied_from_pdf_annotated.txt'
with open(filename, encoding='utf8') as f:
    raw_text = f.read()
print(len(raw_text))
print(raw_text[:100])
```

```python
result = transpose_text(decode_every, raw_text)
print(len(result))
print(result[:100])
```

Now the Russian code-switching can be restored in the file

```python
legacy_to_unicode = make_backwards(decode_every)
print(len(legacy_to_unicode))
```

```python
def restore_cyrillic(text, legacy_to_unicode=legacy_to_unicode):
    borders_span = re.compile(r'%%%%%(.*?)%%%%%', flags=re.S)
    # exclude punctuation from backward replacement
    mapping = {k: v for k, v in legacy_to_unicode.items() if v.isalpha()}
    return borders_span.sub(lambda x: transpose_text(mapping, x.group(1)), text)
```

```python
result_new = restore_cyrillic(result)
with open(filename[:-4] + '_decoded.txt', 'w', encoding='utf8') as f:
    f.write(result_new)
print(len(result_new))
print(result_new[:100])
```

```python
check_OOV_symbols(legacy_to_unicode, result_new)
```

### Environment setup

```python
# !pip install pytesseract
# !pip install pdf2image
# !pip install Pillow

import pytesseract
from pdf2image import convert_from_path
from PIL import Image
from tqdm import tqdm
```

1. We need tesseract files: download from https://github.com/UB-Mannheim/tesseract/wiki, save to "Program Files", add to PATH (to access PATH use Windows+R > sysdm.cpl > Advanced > Environment variables, then use "Create")
2. We need specific trained model: download from https://github.com/calfa-co/hye-tesseract, save to Tesseract '.../tessdata'

A guide to rely on: https://towardsdatascience.com/extract-text-from-memes-with-python-opencv-tesseract-ocr-63c2ccd72b69/

3. To work with PDF files, we also need poppler for Windows: download from https://github.com/oschwartz10612/poppler-windows/releases/, add 'poppler-.../bin' to PATH

A guide to rely on: https://www.geeksforgeeks.org/python/convert-pdf-to-image-using-python/

### PDF files OCR

```python
filepath = '../Тексты для корпуса/Патканов 1875 донской.pdf'
images = convert_from_path(filepath, dpi=300)
print(len(images))

with open(filepath[:-4] + '.txt', 'a', encoding='utf-8') as f:
    for i in tqdm(range(len(images))):
        text = pytesseract.image_to_string(images[i], lang='hye-calfa-n', config='--dpi 300')
#         print(text)
        f.write(f'=== {filepath} page_{i + 1} ===\n')
        f.write(text)
        f.write('\n')

#     with open(f'page_{i + 1}.txt', 'w', encoding='utf-8') as f:
#         f.write(text)
#     images[i].save('page'+ str(i) +'.jpg', 'JPEG')
```

### Get token counts for Nor-Nakhichevan texts

```python
from collections import Counter
from itertools import chain
from pprint import pprint
```

```python
def get_token_counts(preprocessed_text):
    return Counter(chain.from_iterable(preprocessed_text))
```

```python
token_counts = get_token_counts(df['preprocessed'])
print(len(token_counts) - 2)
print()
pprint([i for i in token_counts.most_common(300) if i[0] != '' and i[0] != '–'])
```

Write token counts to file

```python
with open('NNA_corpus_wordlist.tsv', 'w', encoding='utf8') as f:
    for i in [i for i in token_counts.most_common() \
              if i[0] != '' and i[0] != '–']:
        f.write(i[0] + '\t' + str(i[1]) + '\n')
```

### Frequency distribution

```python
df = pd.read_csv('NNA_corpus_wordlist.tsv', sep='\t', \
                 header=None, names=['token', 'count'])
df.head(3)
```

```python
print(len(df['count'].value_counts()))
df['count'].value_counts().head(20)
# most tokens are very infrequent
```

```python
df['count'].plot()
```

### Get word per million (WPM) for NNA tokens

```python
text_size = 102592
token_wpm = sorted({
    k: round(v / text_size * 1000000) for k, v in token_counts.items() \
    if k != '' and k != '–'}.items(), key=lambda x: x[1], reverse=True)
print()
pprint(token_wpm[:300])
```
